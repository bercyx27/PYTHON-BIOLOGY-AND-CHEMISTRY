## Reading a FASTA file

Write a parser thats reads and processes a FASTA file called `ecoli_sequences.fasta` into a list named `ecoli`. Repeat the process for the files `archaea_sequences.fasta` and `plant_sequences.fasta`. Make sure to name the resulting variables appropriately to ensure clean code principles.

**FASTA Schema**

```
0 | >Header1
1 | Sequence1
2 | >Header2
3 | Sequence2
```

````python
def parse_fasta(filepath):
    """
    Reads a FASTA file and extracts the headers and sequences.
    Returns a list of dictionaries.
    """
    sequences_list = []

    # Open the file safely using 'with'
    with open(filepath, 'r') as file:
        current_header = ""
        current_sequence = ""

        for line in file:
            line = line.strip() # Remove leading/trailing whitespace and newlines

            # Skip empty lines
            if not line:
                continue

            # Check if the line is a header
            if line.startswith(">"):
                # If we already have a header stored, save the previous record
                # before starting this new one.
                if current_header:
                    sequences_list.append({
                        "header": current_header,
                        "sequence": current_sequence
                    })

                # Store the new header (removing the '>' character)
                current_header = line[1:]
                # Reset the sequence string for the new record
                current_sequence = ""
            else:
                # If it's not a header, it's sequence data.
                # We append it in case a sequence spans multiple lines.
                current_sequence += line

        # Don't forget to append the very last record in the file once the loop finishes!
        if current_header:
            sequences_list.append({
                "header": current_header,
                "sequence": current_sequence
            })

    return sequences_list

# Process the files and assign them to cleanly named variables
ecoli = parse_fasta("ecoli_sequences.fasta")
archaea = parse_fasta("archaea_sequences.fasta")
plants = parse_fasta("plant_sequences.fasta")

# Example of how to inspect the results
print(f"Loaded {len(ecoli)} sequences for E. coli.")
````

## Writing of a FASTA file

Write all three lists that you've just created using your parser to a new FASTA file called `all_sequences.fasta`. When merging your data into a single file, make sure to assign a header for each sequence that includes information about the organism and an appropriate identifier. Be careful for duplicate headers, each one should be unique.

````python
def write_combined_fasta(datasets, output_filepath):
    """
    Takes a dictionary of sequence lists and writes them to a single FASTA file.
    Automatically generates unique headers using the organism name and an index.
    """
    # Using a context manager to safely open and automatically close the file
    with open(output_filepath, 'w') as file:

        # Loop through each organism and its corresponding list of sequences
        for organism, sequences in datasets.items():

            # enumerate(..., start=1) gives us a running counter (1, 2, 3...) to ensure uniqueness
            for index, record in enumerate(sequences, start=1):

                # 1. Clean the original header (replace spaces with underscores to keep FASTA strict)
                clean_original = record['header'].replace(' ', '_')

                # 2. Use string formatting to create the new, unique header
                # Schema: >Organism_Number_OriginalHeader
                new_header = f">{organism}_{index}_{clean_original}"

                # 3. Write the header and sequence to the file
                # \n ensures the next entry goes on a new line
                file.write(f"{new_header}\n")
                file.write(f"{record['sequence']}\n")

# Combine the lists into a dictionary so we can pass organism names alongside the data
datasets_to_merge = {
    "Ecoli": ecoli,
    "Archaea": archaea,
    "Plant": plants
}

# Execute the function
write_combined_fasta(datasets_to_merge, "all_sequences.fasta")

print("File 'all_sequences.fasta' has been successfully created!")
````
