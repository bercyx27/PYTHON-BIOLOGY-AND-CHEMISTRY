## Determination of sequences lengths

Load the FASTA file `all_sequences.fasta` by using your own FASTA reader or `biopython` from the last exercise. Save all sequences in a list called `sequences` for the following exercises. Now determine the length of all sequences and store the result in a new list called `sequence_lengths`. Finally, print the lowest and highest value in this list.

```python
# 1. Load the FASTA file using a custom reader
sequences = []
current_sequence = []

with open("all_sequences.fasta", "r") as file:
    for line in file:
        line = line.strip()
        # Skip empty lines
        if not line:
            continue

        if line.startswith(">"):
            # If we already have a sequence built up, save it to our list
            if current_sequence:
                sequences.append("".join(current_sequence))
                current_sequence = [] # Reset for the next sequence
        else:
            # Add the sequence string to our current block
            current_sequence.append(line)

    # Don't forget to append the very last sequence in the file
    if current_sequence:
        sequences.append("".join(current_sequence))

# 2. Determine the length of all sequences
sequence_lengths = [len(seq) for seq in sequences]

# 3. Print the lowest and highest values
if sequence_lengths:
    print(f"Lowest sequence length: {min(sequence_lengths)}")
    print(f"Highest sequence length: {max(sequence_lengths)}")
else:
    print("The FASTA file is empty or no sequences were found.")
```

## Calculation of the GC-Content

Now calculate the GC-Content for each sequence by counting `G` and `C` respectively and divide the sum of both numbers by the sequence length. Store the result for each sequence in a list called `gc_content` and print the highest and lowest value of that list.

**Tips**

> * Each string has internal functions that for instance can be used to count specific characters. Make use of it!

```python
# Assuming 'sequences' is the list of sequence strings from the previous exercise

# 1. Initialize the list to store the results
gc_contents = []

# 2. Calculate GC content for each sequence
for seq in sequences:
    # Ensure the sequence is uppercase so we don't miss lowercase 'g' or 'c'
    seq_upper = seq.upper()

    g_count = seq_upper.count('G')
    c_count = seq_upper.count('C')
    seq_length = len(seq_upper)

    # Avoid division by zero if an empty sequence slipped in
    if seq_length > 0:
        gc_content = (g_count + c_count) / seq_length
        gc_contents.append(gc_content)
    else:
        gc_contents.append(0.0) # Or None, depending on how you want to handle errors

# 3. Print the lowest and highest values
if gc_contents:
    # Formatting to 4 decimal places for cleaner output
    print(f"Lowest GC content:  {min(gc_contents):.4f} ({min(gc_contents)*100:.2f}%)")
    print(f"Highest GC content: {max(gc_contents):.4f} ({max(gc_contents)*100:.2f}%)")
else:
    print("The gc_contents list is empty.")
```

## Translation of a DNA sequence

Translate each sequence into its respective amino acid sequence. In order to do that, start by splitting the sequences into triplets. Now use these triplets and construct a new sequence by assigning the corrsponding amino acid. Store the result in a new list called `proteins` and print a single sequence of your choice.

There is a helper function `to_triplets` that will take care of splitting a gene into its respective triplets. In order to convert the triplets, use the dictionary `CODON_TABLE` which maps from triplet to amino acid.

**Tips**

> * Initially you have downloaded the `utils.py` file which contains both the function and codon table. Inspect the content of the file to read the docstring of the `to_triplets` function and the structure of `CODON_TABLE`.

```python
from utils import to_triplets, CODON_TABLE
```

```python
# Import the helper function and codon table from your utils.py file
from utils import to_triplets, CODON_TABLE

# Assuming 'sequences' is the list of DNA sequences from the previous exercise
proteins = []

# 1. Iterate over each DNA sequence
for seq in sequences:
    # 2. Split the sequence into a list of triplets (codons)
    triplets = to_triplets(seq)

    # 3. Translate triplets to amino acids
    # We use a list comprehension and the CODON_TABLE dictionary.
    # .get(codon, '?') is used safely in case an invalid/partial triplet exists.
    amino_acids = [CODON_TABLE.get(codon, '?') for codon in triplets]

    # Join the individual amino acid letters back into a single string
    protein_sequence = "".join(amino_acids)

    # Store in the proteins list
    proteins.append(protein_sequence)

# 4. Print a single sequence of your choice (e.g., the very first one)
if proteins:
    print("First translated protein sequence:")
    print(proteins[0])
else:
    print("No sequences to translate.")
```
